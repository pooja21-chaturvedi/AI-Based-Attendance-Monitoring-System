## 2. Face detection on an image
Put a photo you have permission to use at `student_images/test.jpg` (this folder is git-ignored, so it won't be uploaded).

```python
from matplotlib import pyplot as plt
import cv2

n, boxes = detect_faces_in_image("../student_images/test.jpg", save_to="../student_images/test_out.jpg")
print("Faces detected:", n)

img = cv2.imread("../student_images/test_out.jpg")
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title(f"Faces detected: {n}")
plt.show()
```

> Detection only finds face-shaped regions; it does **not** identify anyone. Recognition is a later, separate step.
